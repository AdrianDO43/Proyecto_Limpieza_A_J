```python
import numpy as np

supply = np.genfromtxt('SupplyPlan_v2.csv', delimiter=',', skip_header=1, filling_values=0))

# 1. np.isnan(arr) - Detectar valores NaN
nan_mask = np.isnan(supply)  # Devuelve un array booleano donde hay NaN
print("¿Dónde hay valores NaN?:", nan_mask)

# 2. np.nanmean(arr) - Calcular la media ignorando NaN
# Calcula la media sin NaN
mean_value = np.nanmean(supply)  
print("Media sin contar los NaN:", mean_value)

# Funciones relacionadas:
# Mediana ignorando NaN
# Desviación estándar ignorando NaN
median_value = np.nanmedian(supply)  
std_value = np.nanstd(supply)  
print("Mediana:", median_value, "| Desviación estándar:", std_value)

# 3. np.where(cond, x, y) - Reemplazar valores NaN por la media
supply_filled = np.where(np.isnan(supply), mean_value, supply)
print("Array con NaN reemplazados por la media:", supply_filled)

# Función relacionada:
# np.nan_to_num() - Reemplaza NaN con 0 (o un valor especificado)
supply_zero_filled = np.nan_to_num(supply, nan=0)
print("Array con NaN reemplazados por 0:", supply_zero_filled)

# 4. arr[~np.isnan(arr)] - Filtrar valores eliminando NaN
supply_clean = supply[~np.isnan(supply)]
print("Array sin valores NaN:", supply_clean)

# Función relacionada:
finite_values = supply[np.isfinite(supply)]  # Similar a eliminar NaN
print("Valores finitos (sin NaN ni infinitos):", finite_values)
```

```python
import pandas as pd

# Crear un DataFrame con valores faltantes
df = pd.read_csv('SupplyPlan_v2.csv')

# Detectar valores NaN en el DataFrame
print("¿Dónde hay valores NaN?")
print(df.isna())  # Devuelve True donde hay NaN y False donde hay datos

# Alternativa: df.notna() devuelve True donde hay valores NO nulos
print("\n¿Dónde NO hay valores NaN?")
print(df.notna())

# Eliminar filas con valores NaN
df_dropped = df.dropna()
print("\nDataFrame después de eliminar filas con NaN:")
print(df_dropped)

# Alternativa: Eliminar columnas con NaN usando axis=1
df_dropped_cols = df.dropna(axis=1)
print("\nDataFrame después de eliminar columnas con NaN:")
print(df_dropped_cols)

# Rellenar NaN con la media de cada columna
df_filled = df.fillna(df.mean())
print("\nDataFrame con NaN reemplazados por la media:")
print(df_filled)

# Alternativa: Rellenar NaN con la mediana
df_filled_median = df.fillna(df.median())
print("\nDataFrame con NaN reemplazados por la mediana:")
print(df_filled_median)

# Alternativa: Rellenar NaN con el valor más frecuente (moda)
df_filled_mode = df.fillna(df.mode().iloc[0])
print("\nDataFrame con NaN reemplazados por la moda:")
print(df_filled_mode)

# Alternativa: Rellenar NaN con el último valor válido (forward fill)
df_ffill = df.fillna(method="ffill")
print("\nDataFrame con forward fill (rellena con el valor anterior):")
print(df_ffill)

# Alternativa: Rellenar NaN con el siguiente valor válido (backward fill)
df_bfill = df.fillna(method="bfill")
print("\nDataFrame con backward fill (rellena con el siguiente valor):")
print(df_bfill)
```
