```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../Sample.Data.v3/Sample Data v3/SupplyPlan_v2.csv")
print("Todas las columnas:", df)
```

```python
array_np = np.array(df["planningCycle"],df["quantity"])
# 1. np.isnan(arr) - Detectar valores NaN
nan_mask = np.isnan(array_np)  # Devuelve un array booleano donde hay NaN
print("¿Dónde hay valores NaN?:", nan_mask)

# 2. np.nanmean(arr) - Calcular la media ignorando NaN
mean_value = np.nanmean(array_np)  # Calcula la media sin NaN
print("Media sin contar los NaN:", mean_value)


# 3. np.nan_to_num() - Reemplaza NaN con 0 (o un valor especificado)
array_np_zero_filled = np.nan_to_num(array_np, nan=0)
print("Array con NaN reemplazados por 0:", array_np_zero_filled)

# 4. arr[~np.isnan(arr)] - Filtrar valores eliminando NaN
array_np_clean = array_np[~np.isnan(array_np)]
print("Array sin valores NaN:", array_np_clean)
```
